```python
text = text = "Aujourd'hui, il fait chaud au soleil ; la mer est belle ! Les touristes visitent la plage et admirent la nature. #Vacances #Nature (Magnifique !)"
```

```python
print(text)
```

```python
text_lower = text.lower()

print(text_lower)
```

```python
import string

text_clean = text_lower.translate(str.maketrans('', '', string.punctuation))

print(text_clean)
```

```python
//partie3
```

```python
from nltk.tokenize import word_tokenize
```

```python
tokens = word_tokenize(text_clean)

print(tokens)
```

```python
from nltk.corpus import stopwords
```

```python
stop_words = set(stopwords.words('english'))

print(stop_words)
```

```python
tokens_filtrés = []

for word in tokens:
    if word not in stop_words:
        tokens_filtrés.append(word)

print(tokens_filtrés)
```

```python
tokens_filtrés = [word for word in tokens if word not in stop_words]

print(tokens_filtrés)
```

```python
stop_words.add("aujourdhui")
```

```python
tokens_filtrés = [word for word in tokens if word not in stop_words]

print(tokens_filtrés)
```

```python
nombre_avant = len(tokens)
nombre_apres = len(tokens_filtrés)

nombre_supprime = nombre_avant - nombre_apres

proportion = (nombre_supprime / nombre_avant) * 100

print("Nombre de mots avant :", nombre_avant)
print("Nombre de mots après :", nombre_apres)
print("Nombre de mots supprimés :", nombre_supprime)
print("Proportion supprimée :", proportion, "%")
```

```python
import spacy

nlp = spacy.load("fr_core_news_sm")

text = "Aujourd'hui, il fait chaud au soleil ; la mer est belle ! Les touristes visitent la plage et admirent la nature. #Vacances #Nature (Magnifique !)"

doc = nlp(text)

tokens = [
    token for token in doc
    if not token.is_punct and not token.is_space
]

tokens_lemmatises = [token.lemma_ for token in tokens]

print("Avant :", [token.text for token in tokens])
print("Après :", tokens_lemmatises)
```
